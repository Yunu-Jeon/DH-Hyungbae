## 파일 3개 병합
- nfm_hyungbae.xlsx
- nfm_dallyeong.xlsx
- nfm_portrait.xlsx

```python
import pandas as pd
import os
import re

DATA_DIR = "../data"

# 1. 한글, 한자 분리해주는 함수
def split_hangeul_hanja(name):
    """예: '흉배(胸背)' -> ('흉배', '胸背') / 괄호 없으면 한자명은 빈칸"""
    m = re.match(r'^(.*?)\(([^)]*)\)\s*$', str(name).strip())
    if m:
        return m.group(1).strip(), m.group(2).strip()
    else:
        return str(name).strip(), ""

# 2. 엑셀 표 제작
def to_team_format(df, category_name, category_code, collector="전윤우"):
    result = pd.DataFrame()

    # default
    result["임시ID"] = [f"YNFM{category_code}{i+1:02d}" for i in range(len(df))]
    result["분류"] = category_name
    result["소장처"] = "국립민속박물관"
    result["소장처유물번호"] = df["소장품 번호"].values
    names = df["소장품 명칭"].apply(split_hangeul_hanja)
    result["한글명"] = [n[0] for n in names]
    result["한자명"] = [n[1] for n in names]
    result["영어명"] = ""
    result["URL"] = df["detail_url"].values
    result["수집자"] = collector

    # metadata
    result["image_url"] = df["image_url"].values
    result["국적/시대"] = df["국적/시대"].values
    result["용도/기능"] = df["용도/기능"].values
    result["내용"] = df["내용"].values
    result["크기"] = df["크기"].values
    result["seq"] = df["seq"].values

    return result

# 2. 파일 불러오기
hyungbae_df  = pd.read_excel(os.path.join(DATA_DIR, "nfm_hyungbae.xlsx"))
dallyeong_df = pd.read_excel(os.path.join(DATA_DIR, "nfm_dallyeong.xlsx"))
portrait_df  = pd.read_excel(os.path.join(DATA_DIR, "nfm_portrait.xlsx"))

# 3. 양식 변환
hyungbae_team  = to_team_format(hyungbae_df,  "흉배", "H")
dallyeong_team = to_team_format(dallyeong_df, "단령", "D")
portrait_team  = to_team_format(portrait_df,  "초상", "P")

# 4. 합치기
final = pd.concat([hyungbae_team, dallyeong_team, portrait_team], ignore_index=True)
print(final.shape)

# 5. 최종 저장
final.to_excel(os.path.join(DATA_DIR, "nfm_merge.xlsx"), index=False)

final.head()
```
