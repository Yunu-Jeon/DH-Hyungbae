## 파일 3개 병합
- nfm_hyungbae.xlsx
- nfm_dallyeong.xlsx
- nfm_portrait.xlsx

```python
import pandas as pd
import os

DATA_DIR = "../data"

# 1. data 폴더 안에서 .xlsx 파일 찾기 (병합 결과 파일은 제외)
file_format = ".xlsx"
file_list = [
    f for f in os.listdir(DATA_DIR)
    if f.endswith(file_format) and f != "nfm_merge.xlsx"
]
print(file_list)

# 2. 하나씩 읽어서 리스트에 모으고 한 번에 합치기
df_list = []
for file in file_list:
    filepath = os.path.join(DATA_DIR, file)   # data 폴더 경로까지 붙여줘야 함
    temp = pd.read_excel(filepath)
    df_list.append(temp)

df = pd.concat(df_list, ignore_index=True)
print(df.shape)

# 3. 병합 결과 저장
df.to_excel(os.path.join(DATA_DIR, "nfm_merge.xlsx"), index=False)
```
