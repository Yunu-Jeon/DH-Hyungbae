## <국립민속박물관 병합 + 팀 양식 변환 + 이미지 파일명 변경>
- 01~03번에서 각각 만든 흉배/단령/초상 결과를 여기서 한 번에 정리
- 병합 → 팀 양식 변환(임시ID 부여) → 이미지 파일명 변경 순서로 진행

```python
import os
import pandas as pd

# utils.py에 미리 정의해둔 함수들 가져오기
from utils import to_team_format, rename_images_by_team_format

DATA_DIR = "../data"
```

### 1. 3개 카테고리 결과 병합
- 01~03번 노트북이 각각 저장해둔 `nfm_hyungbae.xlsx`, `nfm_dallyeong.xlsx`, `nfm_portrait.xlsx`를 읽어와서 합침
- 이 병합본(`nfm_merge.xlsx`)은 각 노트북에서 뽑은 원본 컬럼 그대로 합쳐둔 것 (참고용)

```python
hyungbae_df = pd.read_excel(os.path.join(DATA_DIR, "nfm_hyungbae.xlsx"))
dallyeong_df = pd.read_excel(os.path.join(DATA_DIR, "nfm_dallyeong.xlsx"))
portrait_df = pd.read_excel(os.path.join(DATA_DIR, "nfm_portrait.xlsx"))

print("흉배:", hyungbae_df.shape, "/ 단령:", dallyeong_df.shape, "/ 초상:", portrait_df.shape)

hyungbae_team = to_team_format(hyungbae_df, "흉배", "H")        # utils.py에 존재
dallyeong_team = to_team_format(dallyeong_df, "단령", "D")
portrait_team = to_team_format(portrait_df, "초상", "P")

merged_df = pd.concat([hyungbae_df, dallyeong_df, portrait_df], ignore_index=True)
team_df = pd.concat([hyungbae_team, dallyeong_team, portrait_team], ignore_index=True)
team_df.to_excel(os.path.join(DATA_DIR, "nfm_merge.xlsx"), index=False)

team_df = pd.concat([hyungbae_team, dallyeong_team, portrait_team], ignore_index=True)
team_df.head()
```

### 2. 이미지 파일명을 임시ID로 변경

    - ex) YNFM0000

```python
rename_images_by_team_format(team_df)
```
