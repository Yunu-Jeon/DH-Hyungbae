### 파일명 변경
- 따로 만든 이유는 이미지 거르는 작업이 필요했기 떄문 (처음부터 고착화를 하게 되면 이미지 파일을 매번 바꿔야 함)

```python
import pandas as pd
import os

team_df = pd.read_excel("../data/국립민속박물관_전윤우.xlsx")

folder_map = {
    "흉배": "hyungbae",
    "단령": "dallyeong",
    "초상": "portrait",
}

renamed = 0
skipped = 0

for _, row in team_df.iterrows():
    category = row["분류"]
    folder_name = folder_map.get(category)
    if folder_name is None:
        print(f"  [건너뜀] 알 수 없는 분류: {category}")
        skipped += 1
        continue

    seq = row["seq"]
    new_id = row["임시ID"]

    old_path = os.path.join("..", "image", folder_name, f"{seq}.jpg")
    new_path = os.path.join("..", "image", folder_name, f"{new_id}.jpg")

    if not os.path.exists(old_path):
        print(f"  [파일 없음] {old_path}")
        skipped += 1
        continue

    os.replace(old_path, new_path)  # 같은 폴더 안에서 이름만 바꿈 (덮어쓰기 허용)
    renamed += 1

print(f"완료! 이름 바꾼 파일: {renamed}개, 건너뛴 파일: {skipped}개")
```
